```python
import yfinance as yf
import pandas as pd

# Download Reliance Industries stock data for the last 6 months
stock = yf.Ticker("RELIANCE.NS")
data = stock.history(period="6mo")

# Show the first 5 rows
print(data.head())

# Save to CSV in your Resources folder
data.to_csv("../Resources/reliance_stock_data.csv")
print("\n✅ Data saved successfully!")
```
